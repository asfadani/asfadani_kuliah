**Coin Change**

Kode ini digunakan untuk mencari jumlah koin minimum yang diperlukan untuk mencapai nominal uang tertentu menggunakan pendekatan greedy.

```python
def coin_change_greedy(amount, coins):
    coins.sort(reverse=True)
    result = []
    for coin in coins:
        while amount >= coin:
            amount -= coin
            result.append(coin)
    return result

```

```python
# Contoh penggunaan
amount = 57
coins = [25, 10, 5, 1]
change = coin_change_greedy(amount, coins)
print("Koin yang digunakan:", change)
```

**Fractional Knapsack**

Masalah Fractional Knapsack diselesaikan dengan mengambil barang berdasarkan rasio nilai terhadap berat (value/weight) tertinggi terlebih dahulu untuk memaksimalkan total nilai.

```python
def fractional_knapsack(items, kapasitas):
    # items = list of tuples (value, weight)
    items = sorted(items, key=lambda x: x[0]/x[1], reverse=True)
    total_value = 0.0
    for value, weight in items :
        if kapasitas >= weight:
            kapasitas -= weight
            total_value += value
        else:
            total_value += value* (kapasitas/weight)
            break
    return total_value
```

```python
# Contoh penggunaan
items = [(60, 10), (100, 20), (120, 30)] # (nilai, berat)
capacity = 50
max_value = fractional_knapsack(items, capacity)
print("Nilai maksimum yang dapat dibawa:", max_value)
```

**Minimum Spanning Tree: Prim**

Algoritma Prim membangun Minimum Spanning Tree (MST) dengan memulai dari satu simpul dan secara bertahap menambahkan sisi berbobot terkecil.

```python
import heapq

def prim_mst(graph, start):
    visited = set()
    min_heap = [(0, start)]
    total_weight = 0

    while min_heap:
        weight, node = heapq.heappop(min_heap)
        if node in visited:
            continue
        visited.add(node)
        total_weight += weight
        for neighbor, edge_weight in graph[node]:
            if neighbor not in visited:
                heapq.heappush(min_heap, (edge_weight, neighbor))
    return total_weight

```

```python
# Representasi graf: adjacency list
graph = {
    'A': [('B', 2), ('C', 3)],
    'B': [('A', 2), ('C', 1), ('D', 1)],
    'C': [('A', 3), ('B', 1), ('D', 4)],
    'D': [('B', 1), ('C', 4)]
}
print("Total bobot MST (Prim):", prim_mst(graph, 'A'))
```

**Minimum Spanning Tree: Kruskal**

Algoritma Kruskal membangun MST dengan mengurutkan semua sisi graf berdasarkan bobot terkecil dan menambahkannya satu per satu asalkan tidak membentuk siklus (menggunakan teknik Union-Find).

```python
def kruskal_mst(edges, n_nodes):
    parent = {i: i for i in range(n_nodes)}

    def find(x):
        while parent[x] != x:
            x = parent[x]
        return x
    
    def union(x, y):
        root_x = find(x)
        root_y= find(y)
        if root_x != root_y:
            parent[root_y] = root_x
            return True
        return False
    
    edges.sort(key=lambda x: x[2])
    total_weight = 0
    for u, v, weight in edges:
        if union(u, v):
            total_weight += weight
    return total_weight
```

```python
# edges: (node1, node2, weight), nodes are represented by integers 
edges = [(0, 1, 2), (0, 2, 3), (1, 2, 1), (1, 3, 1), (2, 3, 4)]
print("Total bobot MST (Kruskal):", kruskal_mst(edges, 4))
```
