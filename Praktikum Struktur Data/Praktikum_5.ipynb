```python
def linear_search(arr, N, x):

    for i in range(0, N):
        if (arr[i] == x):
            return i
    return -1


def binnary_search(arr, l, r, x):

    while l <= r:
        mid = l + (r-l) // 2

        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            l = mid + 1
        else:
            r = mid + 1
    return -1


def rekrusif_binnary(arr, l, r, x):

    if r >= l:
        mid = l + (r-l) // 2

        if arr[mid] == x:
            return mid
        elif arr[mid] > x:
            return rekrusif_binnary(arr, l, mid-1, x)
        else:
            return rekrusif_binnary(arr, mid+1, r, x)
        
    else:
        return -1


def jump_search(arr, x, n):
    import math
    step = math.sqrt(n)

    prev = 0
    while arr[int(min(step, n)-1)] < x:
        prev = step
        step += math.sqrt(n)
        if prev >= n:
            return -1
        
    while arr[int(prev)] < x:
        prev += 1
        if prev == min(step, n):
            return -1
        
    if arr[int(prev)] == x:
        return prev
    
    return -1
```

Linear Search

```python
arr = [20, 10, 30, 50, 40]
x = 10
N = len(arr)

hasil = linear_search(arr, N, x)
if(hasil == -1):
    print("Element yang dicari tidak ada di dalam array")
else:
    print(f"Element yang dicari berada di indeks ke -{hasil}")
    
```

Binarry Search Iteratif

```python
arrei = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
x = 72

res = binnary_search(arrei, 0, len(arrei)-1, x)
if(res != -1):
    print(f"Element yang dicari berada di indeks ke -{res}")
else:
    print("Element yang dicari tidak ada di dalam array")
```

Binnary Search Rekrusif

```python
arrbin = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
x = 12

result = rekrusif_binnary(arrbin, 0, len(arrbin)-1, x)
if(result != -1):
    print(f"Element yang dicari berada di indeks ke -{result}")
else:
    print("Element yang dicari tidak ada di dalam array")
```
