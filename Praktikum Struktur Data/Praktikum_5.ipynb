```python
def linear_search(arr, N, x):

    for i in range(0, N):
        if (arr[i] == x):
            return i
    return -1


def binnary_search(arr, l, r, x):

    while l <= r:
        mid = l + (r-l) // 2

        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            l = mid + 1
        else:
            r = mid + 1
    return -1


def rekrusif_binnary(arr, l, r, x):

    if r >= l:
        mid = l + (r-l) // 2

        if arr[mid] == x:
            return mid
        elif arr[mid] > x:
            return rekrusif_binnary(arr, l, mid-1, x)
        else:
            return rekrusif_binnary(arr, mid+1, r, x)
        
    else:
        return -1


def jump_search(arr, x, n):
    import math
    step = math.sqrt(n)

    prev = 0
    while arr[int(min(step, n)-1)] < x:
        prev = step
        step += math.sqrt(n)
        if prev >= n:
            return -1
        
    while arr[int(prev)] < x:
        prev += 1
        if prev == min(step, n):
            return -1
        
    if arr[int(prev)] == x:
        return prev
    
    return -1
```

**Penyelesaian:**
Permasalahan pencarian nilai maksimum dan minimum dari sebuah array dapat menggunakan Algoritma Pencarian Linear:
Dimulai dengan menginisialisasi nilai min dan max sebagai nilai minimum dan maksimum dari dua elemen pertama. Mulai dari elemen ke-3, bandingkan setiap elemen dengan max dan min, dan ubah max dan min sesuai (yaitu, jika elemennya lebih kecil dari min, maka ubah min, jika elemennya lebih besar dari max, maka ubah max, jika tidak, abaikan elemen tersebut).

```python
class nilai:
    def __init__(self):
        self.min = 0
        self.max = 0

def CariMinMax(arr: list, n: int):
    minmax = nilai()

    if n == 1:
        minmax.min = arr[0]
        minmax.max = arr[0]
        return minmax

    if arr[0] > arr[1]:
        minmax.min = arr[1]
        minmax.max = arr[0]
    else:
        minmax.min = arr[0]
        minmax.max = arr[1]

    for i in range(2, n):
        if arr[i] > minmax.max:
            minmax.max = arr[i]
        elif arr[i] < minmax.min:
            minmax.min = arr[i]
    
    return minmax
```

```python
if __name__ == "__main__":
    import random
    arr = []
    for i in range(0, 10):
        i = random.randint(50, 100)
        arr.append(i)
    
    minmax = CariMinMax(arr, len(arr))
    print(f"""
    Array:
    {arr}
    Nilai min nya: {minmax.min}
    Nilai max nya: {minmax.max}
    """)
```
