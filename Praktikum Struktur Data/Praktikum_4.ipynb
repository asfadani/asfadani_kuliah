**QUEUE DALAM PYTHON**

1. `Queue dengan List`

```python
# membuat list queue nya
queue = []

# fungsi append() untuk memaasukkan item ke liat
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

print(f"Menyiapkan antrean\n{queue}")

# fungsi pop(0) untuk mengambil item di awal list
print("\nMengeluarkan item dari antrean")
print(queue.pop(0))
print(queue.pop(0))
print(queue.pop(0))

print(f"\nAntrean setelah menghapus elemen\n{queue}")
```

2. `Queue dengan Collections.deque`

```python
# Modul python untuk membuat antrean dengan collections.deque
from collections import deque

# meyiapkan antreean
queued = deque()

# fungsi append() untuk memasukkan item ke antrean
queued.append(6) 
queued.append(7) 
queued.append(8) 
queued.append(9) 
queued.append(10) 

print(f"Menyiapkan antrean\n{queued}")

# fungsi leftpop() untuk mengambil item di awal list
print("\nMengeluarkan item dari antrean")
print(queued.popleft())
print(queued.popleft())
print(queued.popleft())

print(f"\nAntrean setelah menghapus elemen\n{queued}")
```

3. `Queue dengan queue.LifoQueue`

```python
# Modul python untuk membuat antrean dengan collections.deque
from queue import Queue

# meyiapkan antreean
queueq = Queue(maxsize=5)

# fungsi qsize() untuk menampilkan ukuran
print(f"Ukuran antrean: {queueq.qsize()}\n")

# fungsi put() untuk memasukkan item ke antrean
queueq.put(1)
queueq.put(3)
queueq.put(5)
queueq.put(7)
queueq.put(9)

print(f"Cek Full: {queueq.full()}")
print(f"Ukuran antrean: {queueq.qsize()}")

# fungsi get() untuk mengambil item di awal list
print("\nMengeluarkan item dari antrean")
print(queueq.get())
print(queueq.get())
print(queueq.get())

print(f"\nApakah kosong: {queueq.empty()}")
print(f"Cek Full: {queueq.full()}")
```

4. `Queue dengan Single Linked List`

```python
class QueueL:
    def __init__(self):
        self.queue = list()

    # Fungsi Tambah(nilai) untuk penambahan item ke antrean
    def Tambah(self, nilai):
        if nilai not in self.queue:
            self.queue.insert(0, nilai)
            return True
        return False
    
    # Fungsi Hapus() untuk menghapus item dari antrean
    def Hapus(self):
        if len(self.queue)>0:
            return self.queue.pop()
        return ("Tidak ada item di antrean!")
    
    # Fungsi Ukuran() untuk mengetahui ukuran antrean
    def Ukuran(self):
        return len(self.queue)
    

if __name__ == "__main__":
    queuel = QueueL()
    print(queuel.Tambah(2)) 
    print(queuel.Tambah(4)) 
    print(queuel.Tambah(6)) 
    print(queuel.Tambah(8)) 
    print(queuel.Tambah(10))
    print(queuel.Ukuran())
    print(queuel.Hapus()) 
    print(queuel.Hapus())
    print(queuel.Ukuran()) 
```
