**1. Representasi Graf dengan Adjacency List**

```python
# Membuat graf dengan adjacency list
graph = {
    'A' : ['B', 'C'],
    'B' : ['A', 'D', 'E'],
    'C' : ['A', 'F'],
    'D' : ['B'],
    'E' : ['B', 'F'],
    'F' : ['C', 'E']
}

# Menampilkan graf
def  print_graph(graf):
    for node in graf:
        print(f"{node} -> {graf[node]}")
```

```python
print_graph(graph)
```

**2. Traversal Graf  menggunakan  BFS**

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    while queue:
        vertex = queue.popleft()
        if vertex not in  visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])
```

```python
bfs(graph, "A")
```

**3. Traversal Graf menggunakan DFS**

```python
def dfs(graph, start, visited=None):
    if visited is None :
        visited = set()

    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)
```

```python
dfs(graph, "A")
```

**4. Menghitung jumlah tetangga dari tiap simpul dalam graf**

```python
def count_neighbor(graph):
    result = {}
    for node in graph:
        result[node] = len(graph[node])
    return result
```

```python
jumlah_tetangga = count_neighbor(graph)
print("="*5,"Jumlah tetangga masing masing node","="*5)
for key, value in  jumlah_tetangga.items():
    print(f"Node {key} = {value}")
```

3. Modifikasi fungsi BFS agar mengembalikan list urutan kunjungan, bukan hanya
mencetak.

```python
from collections import deque

def bfs2(graph, start):
    visited = set()
    queue = deque([start])
    urutan = []

    while queue:
        vertex = queue.popleft()
        if vertex not in  visited:
            urutan.append(vertex)
            visited.add(vertex)
            queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])
    return urutan
```

```python
hasil = bfs2(graph_tugas, "C")

print("Urutan Kunjungan (C):")
print(hasil)
```

4. Buat fungsi find_path(graph, start, end) yang mengembalikan satu jalur dari start
ke end jika ada, menggunakan DFS.

```python
def find_path(graph, start, end, path=[]):
    path = path + [start] 

    if start == end :
        return path

    for neighbor in graph[start]:
        if neighbor not in path:
            new_path = find_path(graph, neighbor, end, path)
            if new_path is not None:
                return new_path
```

```python
result = find_path(graph_tugas, "A", "C")

print("Jalur kunjungan (A - C):")
print(result)
```

5. Tambahkan fungsi is_connected(graph) untuk mengecek apakah semua simpul
dalam graf saling terhubung (gunakan kombinasi DFS + logika).

```python
def is_connected(graph):
    starts = []
    for keys in graph:
        starts.append(keys)
    terkunjungi = bfs2(graph, starts[0])
    if len(terkunjungi) == len(starts):
        return True
    return False
```
