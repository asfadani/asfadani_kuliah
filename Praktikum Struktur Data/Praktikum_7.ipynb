**1. Membuat Binary Tree Secara Manual**

Langkah awal adalah membuat tree dengan struktur dasar Node,

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
        
```

lalu menyusun tree secara manual tanpa aturan BST.

```python
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
```

membuat fungsi `inorder` untuk menelusuri tree dari node terkiri, kemudian induknya, lalu anak kanan. Output dari fungsi ini akan mencetak urutan node berdasarkan struktur tersebut.

```python
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)
```

```python
print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

**2. Membuat Binary Search Tree**

Membangun tree secara otomatis berdasarkan aturan BST, di mana data lebih kecil dari root akan diletakkan di kiri, dan data lebih besar di kanan.

```python
class NodeBST:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
        
```

```python
class BinnarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, root, data):
        if root is None:
            return NodeBST(data)
        if data < root.data:
            root.left = self.insert(root.left, data)
        else:
            root.right = self.insert(root.right, data)
        return root
    
    @staticmethod
    def inorderBST(node):
        if node:
            BinnarySearchTree.inorderBST(node.left)
            print(node.data, end=' ')
            BinnarySearchTree.inorderBST(node.right)

    @staticmethod
    def preorderBST(node):
        if node:
            print(node.data, end=' ')
            BinnarySearchTree.preorderBST(node.left)
            BinnarySearchTree.preorderBST(node.right)

    @staticmethod
    def postorderBST(node):
            BinnarySearchTree.preorderBST(node.left)
            BinnarySearchTree.preorderBST(node.right)
            print(node.data, end=' ')

    @staticmethod
    def searchBST(node, key):
        if node is None or node.data == key:
            return node
        if key < node.data:
           return BinnarySearchTree.searchBST(node.left, key)
        return BinnarySearchTree.searchBST(node.right, key)

```

```python
bst = BinnarySearchTree()

rootBST = None
data_node = [50, 30, 70, 20, 40, 60, 80]
for nilai in data_node:
    rootBST = bst.insert(rootBST, nilai)
```

**3. Traversal pada BST**

Tiga jenis traversal digunakan untuk menelusuri BST:
1. Inorder: menampilkan data dari terkecil ke terbesar.
2. Preorder: berguna untuk mencetak struktur tree dari root ke bawah.
3. Postorder: mencetak node setelah seluruh subtree-nya selesai diproses.

```python
print("Inorder Traversal:")
bst.inorderBST(rootBST)

print("\nPreorder Traversal:")
bst.preorderBST(rootBST)

print("\nPostorder Traversal:")
bst.postorderBST(rootBST)
```

**4. Pencarian Nilai dalam BST**

Mencari elemen dalam BST menggunakan metode rekursif. Algoritma bekerja dengan membandingkan nilai target dengan nilai node saat ini. Jika sama, node dikembalikan. Jika lebih kecil, pencarian dilanjutkan ke subtree kiri; jika lebih besar, ke subtree kanan. Program akan mencetak apakah data ditemukan atau tidak.

```python
# Uji pencarian
key = 60
hasil = bst.searchBST(rootBST, key)
if hasil:
    print(f"Key-{key} ditemukan di dalam tree.")
else:
    print(f"Key-{key} tidak ditemukan.")

key = 25
hasil = bst.searchBST(rootBST, key)
if hasil:
    print(f"\nKey-{key} ditemukan di dalam tree.")
else:
    print(f"\nKey-{key} tidak ditemukan.")
```
