**STACK DALAM PYTHON**

Stack dengan list

```python
stack = []
print(f"stack awal :\n{stack}\n")

# operasi append
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

print(f"Stack setelah diappend :\n{stack}\n")

print("operasi pop")
print(stack.pop())
print(stack.pop())
print(stack.pop())

print(f"\nstack setelah dipop: \n{stack}")
```

Stack dengan Collections deque

```python
from collections import deque

stackdeq = deque()
print(f"stack: \n{stackdeq}\n")

# append
stackdeq.append(1)
stackdeq.append(3)
stackdeq.append(3)
stackdeq.append(4)
stackdeq.append(5)

print(f"stack setelah append: \n{stackdeq}\n")

# pop
print("operasi pop")
print(stackdeq.pop())
print(stackdeq.pop())
print(stackdeq.pop())

print(f"\nstack setelah dipop: \n{stackdeq}")
```

Stack dengan queue.LifoQueue

```python
from queue import LifoQueue

LQ = LifoQueue(maxsize=5)

print(f"Isi stack: {LQ.qsize()}\n")

# put
LQ.put(1)
LQ.put(2)
LQ.put(3)
LQ.put(4)
LQ.put(5)

print("Setelah dilakukan put()")
print(f"Full: {LQ.full()}")
print(f"Isi stack: {LQ.qsize()}\n")

# get()
print("operasi get()")
print(LQ.get())
print(LQ.get())
print(LQ.get())


print(f"\nKosong: {LQ.empty()}")
print(f"Isi stack: {LQ.qsize()}\n")
```

Stack dengan Single Linked List

```python
class Node:
    def __init__(self, nilai):
        self.nilai = nilai
        self.lanjut = None

class Stack:

    # buat stack dengan dummy node, dimana lebih mudh mengatasi error
    def __init__(self):
        self.head = Node("Head")
        self.ukuran = 0

    # representasi string pada stack
    def __str__(self):
        saat_ini = self.head.lanjut
        keluar = ""
        while saat_ini:
            keluar += str(saat_ini.nilai) + "->"
            saat_ini = saat_ini.lanjut
        return keluar[:-2]
    
    # mmengambil ukuran saat ini dari stack
    def ambilUkuran(self):
        return self.ukuran
    
    # cek apakah stack kosong
    def cekIsi(self):
        return self.ukuran == 0
    
    # ambil item paling atas dari stack
    def ambil(self):
        # cek apakah kita mengambil stack yang kosong
        if self.cekIsi():
            raise Exception("mengambil dari stack yang kosong")
        return self.head.lanjut.nilai
    
    # memasukkan item ke stack
    def masukkan(self, nilai):
        node = Node(nilai)
        node.lanjut = self.head.lanjut
        self.head.lanjut = node
        self.ukuran += 1

    #  hapus nilai dari stack dan kembalikan
    def hapus(self):
        if self.cekIsi():
            raise Exception("Menghapus dari satck yang kosong")
        dihapus = self.head.lanjut
        self.head.lanjut = self.head.lanjut.lanjut
        self.ukuran -= 1
        return dihapus.nilai
    

# Jalankan kode
if __name__ == "__main__":
    stackL = Stack()
    for i in range(1, 11):
        stackL.masukkan(i)
    print(f"stack: {stackL}")

    for _ in range(1, 6):
        hapus = stackL.hapus()
        print(f"pop: {hapus}")
    print(f"stack: {stackL}")
```
