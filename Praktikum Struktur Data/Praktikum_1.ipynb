```python
a = arr.array('i', [1, 2 ,3])
print(a)
for i in range(0, 3):
  print(a[i], end=" ")
```

```python
b = arr.array('d', [1.5, 2.5, 3.5])
print(b)
for i in range(0, 3):
  print(b[i], end=" ")
```

Operasi array

    1. Menghitung jumlah elemen pada array

```python
c = arr.array('i', [1, 2 ,3, 1, 4, 2, 1])

hitung_1 = c.count(1)
print(c)
print(f"Jumlah elemen (1) di array : {hitung_1}")
```

    2. Membalik elemen

```python
my_array = arr.array('i', [1, 2, 3])
print(f"Sebelum dibalik: {my_array}")
balik = my_array.reverse()
print(f"\nSesudah dibalik: {my_array}")
```

    3. Extending array

```python
d = arr.array('i', [1, 2, 3])
print(f"Sebelum: {d}")
d.extend([4, 5, 6])
print(f"\nSesudah: {d}")
```
