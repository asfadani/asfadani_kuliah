**1. Implementasi Hash Table Sederhana**

```python
class SimpleHashTable:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size
        
    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        self.table[index] = value

    def search(self, key):
        index = self.hash_function(key)
        return self.table[index]
```

```python
# contoh penggunaan
ht = SimpleHashTable(10)
ht.insert("mangga", 100)
ht.insert("jambu", 250)

print(f"Nilai untuk 'mangga': {ht.search("mangga")}")
print(f"Nilai untuk 'jambu': {ht.search("jambu")}")
```

**2. Penggunaan Collision dengan Linear Probing**

```python
class LinearProbingHT:
    def __init__(self, size):
        self.size = size
        self.table = [None] * size
        
    def hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                break
            index = (index + 1) % self.size
            if index == original_index:
                raise("Hash table penuh")
            
        self.table[index] = (key, value)

    def search(self, key):
        index = self.hash_function(key)
        original_index = index

        while self.table[index] is not None:
            if self.table[index][0] == key:
                return self.table[index][1]
            index = (index + 1) % self.size
            if index == original_index:
                break

        return None
```

```python
lpht = LinearProbingHT(5)
lpht.insert("apel", 100)
lpht.insert("pisang", 200)
lpht.insert("melon", 300)

print(f"Nilai untuk 'apel': {lpht.search("apel")}")
print(f"Nilai untuk 'pisang': {lpht.search("pisang")}")
print(f"Nilai untuk melon': {lpht.search("melon")}")
```

**3. Penanganan Collision dengan Chainning**

```python
class ChainingHT():
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]
        
    def hash_function(self, key):
        return hash(key) % self.size
    
    def insert(self, key, value):
        index = self.hash_function(key)
        for i, (k,v) in enumerate(self.table[index]):
            if k == key:
                self.table[index][i] = (key, value)
                return
        self.table[index].append((key, value))

    def search(self, key):
        index = self.hash_function(key)
        for k, v in self.table[index]:
            if k == key:
                return v
        return None
```

```python
cht = ChainingHT(5)
cht.insert("apel", 100)
cht.insert("pisang", 200)
cht.insert("melon", 300)

print(f"Nilai untuk 'apel': {cht.search("apel")}")
print(f"Nilai untuk 'pisang': {cht.search("pisang")}")
print(f"Nilai untuk melon': {cht.search("melon")}")
```

**LATIHAN/TUGAS**

**1. Eksperimen sederhana linear probing**

```python
# a. membuat objek linear probing hash table dengan ukuran 5
lp_tugas = LinearProbingHT(5)

# b. memasukkan data 
data = {
    "A" : 10,
    "B" : 20,
    "C" : 30,
    "D" : 40,
    "E" : 50
}
for key, value in data.items():
    lp_tugas.insert(key, value)

# c. mencetak isi array
for key in data:
    print(f"Nilai untuk {key}: {lp_tugas.search(key)}")
```

**2. Percobaan dasar chaining**

```python
# a. membuat chaining hash table dengan ukuran 3
ch_tugas = ChainingHT(3)

# b. mengisi dengan key dan value nya
keyva = {
    "apel" : 100,
    "melon" : 200,
    "lemon" : 300,
    "pisang" : 400
}
for key, value in keyva.items():
    ch_tugas.insert(key, value)

# c. menampilkan hash table
for key in keyva:
    print(f"Nilai untuk {key}: {ch_tugas.search(key)}")
```

d. penanganan collision pada chaining ini dengan membiarkan key dengan nilai hash yang sama bertumpukan value nya. kenapa bisa begitu karena set up table nya disini menggunakan nested list, dan saat proses memasukkan data jika ada yang sama maka akan di tambahkan ke list nya
