Membuat Linked list

```python
class Node:
    def __init__ (self, dataval=None):
        self.dataval = dataval
        self.nextval = None
```

```python
n1 = Node("Januari")
n2 = Node("Februari")
n3 = Node("Maret")

print(n1.dataval)
print(n2.dataval)
print(n3.dataval)
```

```python
class Linkedlist:
    def __init__(self):
        self.headval = None

    # menelusuri noe dalam linked list
    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval
    
    # penyisipan node di awal
    def TambahAwal(self, newdata):
        NewNode =  Node(newdata)
        NewNode.nextval = self.headval 
        self.headval = NewNode

    # penyisipan node di tengah
    def TambahTengah(self, mid_node, newdata):
        if mid_node is None:
            print("The mentioned node is absent")
            return
        NewNode = Node(newdata)
        NewNode.nextval = mid_node.nextval
        mid_node.nextval = NewNode

    # penyisipan node di tengah
    def TambahAkhir(self, newdata):
        NewNode = Node(newdata)
        if self.headval is None:
            self.headval = NewNode
            return
        last = self.headval
        while(last.nextval):
            last = last.nextval
        last.nextval = NewNode

    # hapus node dengan key
    def RemoveNode(self, Removekey):
        Headval = self.headval
        if (Headval is not None):
            if (Headval.dataval == Removekey):
                self.headval = Headval.next
                Headval = None
                return
        while (Headval is not None):
            if Headval.dataval == Removekey:
                break
            prev = Headval
            Headval = Headval.nextval
        if (Headval == None):
            return
        prev.nextval = Headval.nextval
        Headval = None

    # hapus di awal node
    def HapusAwal(self):
        afterhead = self.headval
        self.headval = afterhead.nextval

    # hapus di akhir node 
    def HapusAkhir(self):
        last = self.headval
        while (last is not None):
            if last.nextval == None:
                break
            prev = last
            last = last.nextval
        if (last == None):
            return
        prev.nextval = last.nextval
        last = None
```

1. Doble linked list

```python
class NodeDouble:
    def __init__ (self, dataval=None):
        self.dataval = dataval
        self.preval = None
        self.nextval = None
```

```python
nd1 = NodeDouble("node 1")
nd2 = NodeDouble("node 2")
nd3 = NodeDouble("node 3")


print(nd1.dataval)
print(nd2.dataval)
print(nd3.dataval)
```

```python
class DobLinkedlist:
    def __init__(self):
        self.headval = None

    def listprint(self):
        printval = self.headval
        while printval is not None:
            print(printval.dataval)
            printval = printval.nextval

    # penyisipan node di tengah
    def TambahTengah(self, mid_node, newdata):
        if mid_node is None:
            print("The mentioned node is absent")
            return
        NewNode = Node(newdata)
        NewNode.nextval = mid_node.nextval
        NewNode.preval = mid_node.preval
        mid_node.nextval = NewNode
        mid_node.nextval.preval = NewNode

    # hapus dengan kata kunci
    def HapusKunci(self, key):
        Headval = self.headval
        if (Headval is not None):
            if (Headval.dataval == key):
                self.headval = Headval.nextval
                if self.headval is not None:
                    Headval.preval = None
                Headval = None
                return
            while (Headval is not None):
                if Headval.dataval == key:
                    break
                prev = Headval
                Headval = Headval.nextval
            if (Headval is None):
                return
            prev.nextval = Headval.nextval
            Headval.nextval.preval = prev
            Headval = None

                
```

```python
dobli = DobLinkedlist()
dobli.headval = nd1
dobli.headval.nextval = nd2
nd2.preval = nd1
nd2.nextval = nd3
nd3.preval = nd2

dobli.listprint()
```

```python
dobli.TambahTengah(nd1, "T. Tengah")
dobli.listprint()
```

```python
na1 = NodeDouble("satu")
na2 = NodeDouble("dua")
na3 = NodeDouble("tiga")
```

```python
dobli = DobLinkedlist()
dobli.headval = na1
dobli.headval.nextval = na2
na2.preval = na1
na2.nextval = na3
na3.preval = na2

dobli.listprint()
```

```python
dobli.HapusKunci("dua")
dobli.listprint()
```

2. Circular Linked list

```python
class NodeCircular:
    def __init__(self, dataval=None):
        self.dataval = dataval
        self.nextval = None
```

```python
nc1 = NodeCircular("pertama")
nc2 = NodeCircular("kedua")
nc3 = NodeCircular("ketiga")

print(nc1.dataval)
print(nc2.dataval)
print(nc3.dataval)
```

```python
class CircularLinked:
    def __init__(self):
        self.headval = None

    def listprint(self):
        printval = self.headval
        while(printval is not None):
            print(printval.dataval)
            printval = printval.nextval
            if printval.nextval == self.headval.nextval:
                break 
        if printval is None:
            return print("tidak ada node tersimpan")  
        print(printval.dataval)
        
        

    # penyisipan akhir
    def TambahAkhirC(self, newdata):
        NewNode = NodeCircular(newdata)
        if self.headval is None:
            self.headval = NewNode
            self.headval.nextval = self.headval
            return
        last = self.headval
        while(last.nextval):
            last = last.nextval
            if last.nextval == self.headval:
                break
        last.nextval = NewNode
        NewNode.nextval = self.headval

    # hapus akhir
    def HapusAkhirC(self):
        last = self.headval
        if last.nextval == self.headval:
            self.headval = None
            return
        while(last is not None):
            if last.nextval == self.headval:
                break
            prev = last
            last = last.nextval
        if (last == None):
            return
        prev.nextval = last.nextval
        last.nextval = self.headval
        last = None


    
```

```python
cr = CircularLinked()
cr.headval = nc1
cr.headval.nextval = nc2
nc2.nextval = nc3
nc3.nextval = nc1

cr.listprint()
```

```python
cr.TambahAkhirC("akhir")

cr.listprint()
```

```python
ncr1 = NodeCircular("Petama")
ncr2 = NodeCircular("Kedua")
ncr3 = NodeCircular("Terakhir")

crl = CircularLinked()
crl.headval = ncr1
crl.headval.nextval = ncr2
ncr2.nextval = ncr3
ncr3.nextval = ncr1

crl.listprint()
```

```python
crl.HapusAkhirC()
crl.listprint()
```
