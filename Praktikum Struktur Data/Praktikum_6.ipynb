`Buble Sort`

```python
def buble_sort(arr):
    n = len(arr)
    for i in range(n-1):
        for j in range(n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1],  arr[j]
    return arr
```

```python
if __name__ == "__main__":
    data = [64, 34, 25, 12, 22, 11, 90]
    sorted_data =  buble_sort(data.copy())
    print(f"""Data sebelum sorting: {data}\nData setelah sorting: {sorted_data}""")
```

**Penjelasan**
1. Fungsi bubble_sort(arr) menerima sebuah daftar (array) sebagai input dan mengurutkannya menggunakan metode Bubble Sort.
2. Dua perulangan for bersarang digunakan untuk membandingkan elemen secara berpasangan dan menukar jika perlu.
3. arr[j] > arr[j + 1] memastikan elemen yang lebih besar berpindah ke posisi yang lebih tinggi, sehingga elemen terbesar "menggelembung" ke posisi terakhirdalam setiap iterasi.
4. Hasil akhir adalah array yang terurut dari terkecil hingga terbesar.

Bubble Sort memiliki implementasi yang sederhana, tetapi tidak efisien untuk jumlah data besar karena kompleksitas waktu O(n2). Selanjutnya, kita akan mempelajari Selection Sort, yang menggunakan pendekatan berbeda denganmencari elemen terkecil dalam setiap iterasi.

`Selection Sort`

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n-1):
        min_index = i
        for j in range(i+1,n):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
    return arr
```

```python
if __name__ == "__main__":
    data = [64, 25, 12, 22, 11]
    sorted_data =  selection_sort(data.copy())
    print(f"""Data sebelum sorting: {data}\nData setelah sorting: {sorted_data}""")
```

**Penjelasan**

1. Fungsi selection_sort(arr) menerima sebuah daftar dan mengurutkannya menggunakan metode Selection Sort.
2. Variabel min_index digunakan untuk menyimpan indeks elemen terkecil dalam iterasi saat ini.
3. Perulangan for j mencari elemen terkecil dari sisa daftar dan memperbarui min_index.
4. Setelah indeks terkecil ditemukan, elemen pada posisi i dan min_index ditukar, sehingga elemen terkecil berada di posisi yang benar.

Selection Sort lebih efisien dibandingkan Bubble Sort dalam hal jumlah pertukaran, tetapi masih memiliki kompleksitas waktu O(n2). Untuk meningkatkan efisiensi, kita bisa menggunakan algoritma Insertion Sort, yang akan kita bahas pada subbab berikutnya.

`Insertion Sort`

```python
def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i-1
        while j >= 0 and arr[j] > key:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr
```

```python
if __name__ == "__main__":
    data = [64, 34, 25, 12, 22, 11, 90]
    sorted_data =  insertion_sort(data.copy())
    print(f"""Data sebelum sorting: {data}\nData setelah sorting: {sorted_data}""")
```

**Penjelasan**

1. Fungsi insertion_sort(arr) menerima daftar dan mengurutkannya menggunakan metode Insertion Sort.
2. Variabel key menyimpan nilai elemen yang akan dimasukkan ke dalam posisi yang benar.
3. Perulangan while digunakan untuk menggeser elemen-elemen yang lebih besar ke kanan sampai ditemukan posisi yang sesuai untuk key.
4. Setelah ditemukan posisi yang benar, nilai key dimasukkan ke dalam daftar.

Insertion Sort lebih efisien dibandingkan Bubble Sort dan Selection Sort untuk dataset
kecil atau hampir terurut. Namun, untuk dataset besar, kita membutuhkan algoritma
yang lebih cepat seperti Merge Sort, yang akan kita bahas pada subbab berikutnya.

`Merge Sort`

```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        l = arr[:mid]
        r = arr[mid:]

        merge_sort(l)
        merge_sort(r)

        i = j = k = 0

        while i < len(l) and j < len(r):
            if l[i] < r[j]:
                arr[k] = l[i]
                i += 1
            else:
                arr[k] = r[j]
                j += 1
                k += 1
        
        while i < len(l):
            arr[k] = l[i]
            i += 1
            k += 1

        while j < len(r):
            arr[k] = r[j]
            j += 1
            k +=1

    return arr
```

```python
if __name__ == "__main__":
    data = [64, 25, 12, 22, 11]
    sorted_data =  merge_sort(data.copy())
    print(f"""Data sebelum sorting: {data}\nData setelah sorting: {sorted_data}""")
```

**Penjelasan**

1. Fungsi merge_sort(arr) membagi daftar menjadi dua bagian (left_half dan right_half) hingga setiap bagian hanya memiliki satu elemen.
2. Rekursi digunakan untuk mengurutkan masing-masing bagian sebelum digabungkan kembali.
3. Proses penggabungan (merging) dilakukan dengan membandingkan elemen- elemen dari kedua bagian dan memasukkannya ke dalam daftar utama (arr).
4. Dua perulangan while terakhir memastikan semua elemen tersisa dimasukkan ke dalam daftar utama jika masih ada yang belum tergabung.

Merge Sort menawarkan efisiensi yang lebih tinggi dibandingkan algoritma sorting
sebelumnya, terutama untuk dataset besar. Namun, ada algoritma lain yang juga
berbasis divide and conquer dan sering digunakan dalam dunia nyata, yaitu Quick Sort,
yang akan kita bahas pada subbab berikutnya.

`Quick Sort`

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) // 2]
    l = [x for x in arr if x < pivot]
    m = [x for x in arr if x == pivot]
    r = [x for x in arr if x > pivot]

    return quick_sort(l) + m + quick_sort(r)
```

```python
if __name__ == "__main__":
    data = [64, 34, 25, 12, 22, 11, 90]
    sorted_data =  quick_sort(data.copy())
    print(f"""Data sebelum sorting: {data}\nData setelah sorting: {sorted_data}""")
```

**Penjelasan**

1. Fungsi quick_sort(arr) bekerja secara rekursif dengan memilih pivot sebagai titik pembanding.
2. Daftar dibagi menjadi tiga bagian:
a.left : elemen yang lebih kecil dari pivot
o middle: elemen yang sama dengan pivot
o right: elemen yang lebih besar dari pivot
3. Proses rekursi dilakukan pada left dan right hingga seluruh elemen terurut.
4. Akhirnya, semua bagian digabungkan kembali menjadi satu daftar yang sudah terurut.

Quick Sort adalah salah satu algoritma sorting tercepat dan sering digunakan dalam
berbagai aplikasi nyata.
